# Model Validation

## Objective
Evaluate rolling out-of-sample VaR exceptions across multiple models and test sensitivity to lookback and decay assumptions.


```python
from src.data import make_demo_prices
from src.portfolio import load_positions_csv, pnl_history_from_prices
from src.backtesting import calibration_sensitivity, compare_backtests

prices = make_demo_prices()
positions = load_positions_csv('../config/portfolio.csv')
pnl = pnl_history_from_prices(prices, positions)['portfolio_pnl']
summary, forecasts = compare_backtests(pnl, ['historical', 'parametric', 'weighted_historical'], window=250, confidence=0.99, decay=0.97)
summary
```

```python
calibration_sensitivity(pnl, windows=[125, 250, 500], confidence=0.99, decays=[0.94, 0.97, 0.99])
```

## Review point
Coverage, independence, and conditional coverage are diagnostics. Parameter sensitivity is reported to expose calibration dependence rather than optimize a model to pass a test.
