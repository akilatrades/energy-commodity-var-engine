## What a signed contract count means

Positive contracts are long. Negative contracts are short. The project uses a fixed position so we can isolate how market-price changes affect P&L.


```python
pnl = pnl_history_from_prices(prices, positions)
pnl.tail()
```

## Core formula

For each futures position:

`daily P&L = contracts × contract multiplier × daily price change`

The portfolio P&L is the sum across positions.
