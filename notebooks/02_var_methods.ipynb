## How to read 99% VaR

A 99% one-day VaR is a model threshold. Under the model and sample, losses are expected to exceed that threshold roughly 1% of days. It is **not** a maximum possible loss.


## Why Expected Shortfall matters

VaR tells you where the tail begins. Expected Shortfall tells you how bad the losses in that tail are on average.
