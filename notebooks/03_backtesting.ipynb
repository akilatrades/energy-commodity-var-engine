# 03 — VaR Backtesting

A risk model should be compared with what actually happened next.


```python
from src.data import make_demo_prices
from src.portfolio import default_energy_book, pnl_history_from_prices
from src.backtesting import rolling_var_forecasts, summarize_backtest

prices = make_demo_prices()
pnl = pnl_history_from_prices(prices, default_energy_book())
results = rolling_var_forecasts(pnl["portfolio_pnl"], window=250, confidence=0.99)
results.tail()
```

```python
summarize_backtest(results, confidence=0.99)
```

## What an exception means

An exception occurs when the realized loss is worse than the VaR forecast. At 99% confidence, a small number of exceptions is normal. Too many, too few, or clustered exceptions can signal calibration or model problems.
