# VaR Methodology Comparison

## Objective
Compare Historical, Parametric Normal, correlated Student-t Monte Carlo, and Weighted Historical risk estimates on the same portfolio.


```python
from src.data import make_demo_prices
from src.portfolio import load_positions_csv, pnl_history_from_prices
from src.var_models import summarize_var_methods

prices = make_demo_prices()
positions = load_positions_csv('../config/portfolio.csv')
pnl = pnl_history_from_prices(prices, positions)
summary = summarize_var_methods(pnl['portfolio_pnl'], prices, positions, confidence=0.99, decay=0.97, n_sims=50000, monte_carlo_df=6)
summary
```

## Review point
Differences across methods should be interpreted through their assumptions rather than ranked mechanically. Historical methods depend on the realized sample, Parametric VaR assumes normality, and the Monte Carlo implementation uses correlated heavy-tailed factor shocks.
