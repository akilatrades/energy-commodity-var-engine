# Stress, Attribution, and Limits

## Objective
Connect the portfolio-level risk number to its position drivers, severe scenarios, and configured limit utilization.


```python
import json
from src.attribution import component_var
from src.data import make_demo_prices
from src.limits import limit_status
from src.portfolio import load_positions_csv, pnl_history_from_prices
from src.stress import historical_replay_scenarios, load_stress_scenarios, run_hypothetical_scenarios
from src.var_models import historical_var

prices = make_demo_prices()
positions = load_positions_csv('../config/portfolio.csv')
pnl = pnl_history_from_prices(prices, positions)
```

```python
component_var(pnl, confidence=0.99)
```

```python
scenarios = load_stress_scenarios('../config/stress_scenarios.csv')
run_hypothetical_scenarios(prices.iloc[-1], positions, scenarios)
```

```python
historical_replay_scenarios(prices, positions, count=5)
```

```python
with open('../config/risk_limits.json') as f:
    limits = json.load(f)
current_var = historical_var(pnl['portfolio_pnl'], 0.99)
limit_status(current_var, limits['var_99_1d_usd'], limits['watch_utilization'])
```
